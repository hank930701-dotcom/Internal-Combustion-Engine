**內燃機（Internal Combustion Engine, ICE）**

* **基本定義與能量轉換**
  * **熱機性質**：將燃料的化學能轉化為軸能（Shaft Work）的熱力裝置。
  * **能量轉換歷程**：燃料與空氣混合燃燒，將**化學能**轉換為**熱能**，使氣缸內氣體的溫度與壓力大幅提高，再利用高壓氣體膨脹推動機構運轉。
  $$E_{\text{chemical}} \rightarrow E_{\text{thermal}} \rightarrow E_{\tau}$$

* **引擎主要分類**
  1. **往復式引擎（Reciprocating Engine）**：具有單一氣缸或多氣缸，甚至多達&nbsp;20&nbsp;個氣缸，大多數內燃機採此形式（如汽、柴油引擎）。
  2. **迴轉式引擎（Rotary Engine）**：如汪克爾引擎（Wankel Engine）。
  3. **噴射引擎（Jet Engine）**
  4. **火箭引擎（Rocket Engine）**
  5. **槍械／管狀火器（Firearms）**：廣義的單次運作內燃機制。

## **1-6內燃機基本引擎循環**


涵蓋四行程與二行程之 **火星塞點火（SI, Spark Ignition）** 與 **壓燃（CI, Compression Ignition）** 引擎的運作原理與比較。


### 一、 四行程引擎 (Four-Stroke Engine)

四行程引擎完成一個工作循環需經過 **曲軸旋轉兩圈（720°）**、**活塞往復四次行程**。

#### 1. 四行程火星塞點火引擎 (Four-Stroke SI Engine)
* **熱力學理想循環**：奧圖循環 (Otto Cycle) - 定容加熱
* **燃料**：汽油 (Gasoline)
* **混合氣形成**：進氣行程中將空氣與汽油的混合氣吸入汽缸（或經由進氣門噴射）。

* **四大行程運作機制**：
  * **進氣行程 (Intake)**：進氣門開啟，活塞由上死點 (TDC) 移至下死點 (BDC)，將**空燃混合氣**吸入汽缸。
  * **壓縮行程 (Compression)**：雙門關閉，活塞由 BDC 移至 TDC，壓縮混合氣。接近 TDC 時，**火星塞點火 (Spark Plug)** 引燃混合氣。
  * **動力行程 (Power / Expansion)**：燃爆產生高溫高壓氣體，推動活塞由 TDC 移至 BDC 作功。
  * **排氣行程 (Exhaust)**：排氣門開啟，活塞由 BDC 移至 TDC，將廢氣排出汽缸。

---

#### 2. 四行程壓燃引擎 (Four-Stroke CI Engine)
* **熱力學理想循環**：柴油循環 (Diesel Cycle) - 定壓加熱 / 混合循環 (Dual Cycle)
* **燃料**：柴油 (Diesel)
* **混合氣形成**：進氣行程僅吸入純空氣，壓縮末期直接將柴油噴入高溫空氣中。

* **四大行程運作機制**：
  * **進氣行程 (Intake)**：進氣門開啟，活塞由 TDC 移至 BDC，僅吸入**純空氣**。
  * **壓縮行程 (Compression)**：雙門關閉，活塞由 BDC 移至 TDC。高壓縮比使空氣達到高溫高壓（超過柴油自燃點）。接近 TDC 時，**噴油嘴 (Fuel Injector)** 噴入高壓柴油。
  * **動力行程 (Power / Expansion)**：柴油自燃爆發，推動活塞由 TDC 移至 BDC 作功。
  * **排氣行程 (Exhaust)**：排氣門開啟，活塞由 BDC 移至 TDC，排出廢氣。

---

### 二、 二行程引擎 (Two-Stroke Engine)

二行程引擎完成一個工作循環只需 **曲軸旋轉一圈（360°）**、**活塞往復兩次行程**。通常無氣門機構，改由活塞掩蓋汽缸壁上的**進氣孔 (Intake Port)**、**掃除孔 (Scavenging Port)** 與 **排氣孔 (Exhaust Port)**。

#### 1. 二行程火花點火引擎 (Two-Stroke SI Engine)
* **結構特點**：利用曲軸箱預壓混合氣，構造簡單、重量輕，但機油易混合燃燒造成污染。

* **兩大行程運作機制**：
  * **向上行程 (Upward Stroke - 壓縮與進氣)**：
    * **活塞上方**：活塞由 BDC 上升掩蓋掃除孔與排氣孔，開始壓縮汽缸內的混合氣。接近 TDC 時火星塞點火。
    * **活塞下方（曲軸箱）**：活塞上升使曲軸箱產生負壓，進氣閥開啟，將**混合氣（含汽油與機油）**吸入曲軸箱。
  * **向下行程 (Downward Stroke - 動力與掃除/排氣)**：
    * **活塞上方**：燃燒膨脹推動活塞下行作功。接近 BDC 時，**排氣孔先開啟**排出廢氣；隨後**掃除孔開啟**，新混合氣進入汽缸並驅趕剩餘廢氣（掃除作用 Scavenging）。
    * **活塞下方（曲軸箱）**：活塞下行壓縮曲軸箱內的混合氣，預備經由掃除孔送入汽缸。

---

#### 2. 二行程壓燃引擎 (Two-Stroke CI Engine)
* **結構特點**：常用於大型船舶或重型機械，通常配備**強制進氣風扇/增壓器 (Blower / Supercharger)** 進行掃除，不依賴曲軸箱預壓。

* **兩大行程運作機制**：
  * **向上行程 (Upward Stroke - 壓縮與掃除)**：
    * 活塞由 BDC 上升掩蓋進氣孔/排氣門，壓縮汽缸內的**純空氣**至極高溫高壓。接近 TDC 時，噴油嘴噴入柴油。
  * **向下行程 (Downward Stroke - 動力與排氣/進氣)**：
    * 柴油自燃推動活塞下行作功。
    * 接近 BDC 時，排氣門（或排氣孔）開啟排出廢氣，同時增壓風扇將**新鮮純空氣**經由進氣孔吹入汽缸完成掃除。

---

### 三、 四種引擎綜合對比表

| 特性 / 引擎類型 | 四行程 SI 引擎 | 四行程 CI 引擎 | 二行程 SI 引擎 | 二行程 CI 引擎 |
| :--- | :--- | :--- | :--- | :--- |
| **點火方式** | 火星塞強制點火 | 壓燃自燃 | 火星塞強制點火 | 壓燃自燃 |
| **進氣內容** | 空氣 + 汽油混合氣 | 純空氣 | 空氣 + 汽油 + 機油 | 純空氣 |
| **主要應用** | 乘用汽機車 | 重型卡車、巴士、工程車 | 小型機車、割草機、鏈鋸 | 大型船舶主機、發電機 |
| **熱效率 / 省油度** | 中等 | 高 (高壓縮比) | 低 (掃氣未燃損失) | 極高 (大型二行程柴油) |
| **排放污染** | 較低 (搭配觸媒) | 較高 (NOx 與懸浮微粒) | 高 (機油燃燒與未燃油氣) | 中等 (需大型廢氣處理) |
| **馬力重量比** | 中等 | 較低 (本體較重) | 高 | 高 (大排氣量下功率大) |
